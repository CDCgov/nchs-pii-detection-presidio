# Apply Text Preprocessing Methods

## Set parameters

```python
eval_subset_data1 = "catalog.schema.table"
eval_subset_data2 = "catalog.schema.table"
# eval_subset_data3= "catalog.schema.table"
# more as needed

target_col = "MY_TARGET"
id_cols = "ID1, ID2"
context_col = "MY_CONTEXT_COL"

drop_punctuation_tb = "catalog.schema.table"
uppercase_tb = "catalog.schema.table"
lowercase_tb = "catalog.schema.table"
context_col_tb = "catalog.schema.table"

access_group = "group-name"
```

## Remove common punctuation

```python
spark.sql(
    "CREATE OR REPLACE TABLE "
    + drop_punctuation_tb
    + """ AS
          (SELECT """
    + id_cols
    + ", "
    + target_col
    + " AS "
    + target_col
    + "_original, REPLACE(REPLACE(REPLACE("
    + target_col
    + """, ',', ''), '"', ''), ':', '') as """
    + target_col
    + """ FROM 
           (SELECT * FROM 
           (SELECT * FROM """
    + eval_subset_data1
    + """
           UNION
           SELECT * FROM """
    + eval_subset_data2
    + """ ) ) )"""
)
```

## All upper case

```python
spark.sql(
    "CREATE OR REPLACE TABLE "
    + uppercase_tb
    + """ AS
          (SELECT """
    + id_cols
    + ", "
    + target_col
    + " AS "
    + target_col
    + "_original, UPPER("
    + target_col
    + """) as """
    + target_col
    + """ FROM 
           (SELECT * FROM 
           (SELECT * FROM """
    + eval_subset_data1
    + """
           UNION
           SELECT * FROM """
    + eval_subset_data2
    + """ ) ) )"""
)
```

## All lower case

```python
spark.sql(
    "CREATE OR REPLACE TABLE "
    + lowercase_tb
    + """ AS
          (SELECT """
    + id_cols
    + ", "
    + target_col
    + " AS "
    + target_col
    + "_original, LOWER("
    + target_col
    + """) as """
    + target_col
    + """ FROM 
           (SELECT * FROM 
           (SELECT * FROM """
    + eval_subset_data1
    + """
           UNION
           SELECT * FROM """
    + eval_subset_data2
    + """ ) ) )"""
)
```

## Add context from another column

```python
spark.sql(
    "CREATE OR REPLACE TABLE "
    + context_col_tb
    + """ AS
          (SELECT """
    + id_cols
    + ", "
    + target_col
    + " AS "
    + target_col
    + "_original, CONCAT(COALESCE("
    + context_col
    + ", ''), ' ', COALESCE("
    + target_col
    + ", '')) as "
    + target_col
    + """ FROM 
           (SELECT * FROM 
           (SELECT * FROM """
    + eval_subset_data1
    + """
           UNION
           SELECT * FROM """
    + eval_subset_data2
    + """ ) ) )"""
)
```

Grant permissions as needed

```python
spark.sql("GRANT SELECT ON " + drop_punctuation_tb + " TO " + access_group)
spark.sql("GRANT SELECT ON " + uppercase_tb + " TO " + access_group)
spark.sql("GRANT SELECT ON " + lowercase_tb + " TO " + access_group)
spark.sql("GRANT SELECT ON " + context_col_tb + " TO " + access_group)
```
